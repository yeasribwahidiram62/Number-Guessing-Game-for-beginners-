```python
import random as rd
def num_guessing_game():
  num_of_chances = 3
  num = rd.randint(1,100)
  guess_count = 0
  while num_of_chances > guess_count:
    user_guess = int(input(("Guess a number between 1 & 100: ")))
    guess_count += 1

    if user_guess == num :
      print("You guessed the number!")
      break
    elif user_guess > num:
      print("Your guess is too high, my friend!")
    elif user_guess < num:
      print("Your guess is too low, my friend!")
    else:
      print("Invalid input!")
  while guess_count == num_of_chances and user_guess != num:
    print("Oh! You lost!")
    break
  print("The number was:", num)
  print("You guessed", guess_count, "times.")
def play ():
  play_again = input("Do you want to play again? (Yes/No): ").lower()
  if play_again == "Yes" or play_again == "yes":
    num_guessing_game()
  else:
    print("Thank you for playing!")



num_guessing_game()
play()
```
